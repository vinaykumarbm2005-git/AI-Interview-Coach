```python
import numpy as np
import pandas as pd

data={"Employee_ID":[1409,2610,2005,1009,1426],
   "Name":["vinay","sujan","vivek","varun","karthik"],
   "Department":["CSE","CSE","EEE","ISE","ISE"],
   "Age":[21,22,24,25,26],
   "Salary":[14,np.nan,18,np.nan,21]
   }
df = pd.DataFrame(data)
print(df)

print(df.isnull().sum())
df["Salary"] = df["Salary"].fillna(
    df.groupby("Department")["Salary"].transform("median")
)
print(df)
```

```python
df["Annual_Bonus"] = df["Salary"] * 0.10
print(df)
average_salary = df.groupby("Department")["Salary"].mean()
print(average_salary)
```

#Pandas Filtering & Conditional Selection


```python
import pandas as pd

data = {
    "Product": ["Laptop", "Phone", "Headphones", "Keyboard", "Mouse"],
    "Category": ["Electronics", "Electronics", "Accessories", "Accessories", "Accessories"],
    "Price": [30000,50000, 2000, 100, 400],
    "Stock": [10, 15, 25, 30, 40]
}
df = pd.DataFrame(data)
print(df)
```

```python
result = df[(df["Price"] > 500) & (df["Stock"] < 50)]

print(result)
```

```python
df["Restock_Needed"] = df["Stock"] < 20

print(df)
```

```python
df = df.sort_values("Price", ascending=False)

print(df)
```
